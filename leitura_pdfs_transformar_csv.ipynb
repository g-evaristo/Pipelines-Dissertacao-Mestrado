```python
# Caminho base dos PDFs
caminho_base = "/content/drive/MyDrive/ARTIGOS SELECIONADOS/0 - Novos Artigos"
pdfs = glob.glob(caminho_base + "/**/*.pdf", recursive=True)

def extrair_texto(pdf_path):
    """Tenta extrair texto de um PDF, retorna None se falhar"""
    try:
        reader = PdfReader(pdf_path)
        texto = ""
        for page in reader.pages:
            try:
                texto += page.extract_text() or ""
            except:
                pass
        return {"arquivo": pdf_path, "texto": texto, "status": "ok" if texto else "erro"}
    except:
        return {"arquivo": pdf_path, "texto": None, "status": "erro"}

# Usa ThreadPoolExecutor para processar vários PDFs simultaneamente
dados = []
with ThreadPoolExecutor(max_workers=8) as executor:  # ajusta workers conforme seu notebook
    futuros = [executor.submit(extrair_texto, p) for p in pdfs]
    for futuro in as_completed(futuros):
        dados.append(futuro.result())
```

```python
# Cria DataFrame
df = pd.DataFrame(dados)
print(df.head())

# Salva em CSV
df.to_csv("/content/extracao_textos_artigos_pdfs.csv", index=False, encoding="utf-8", errors="ignore")

from google.colab import files
files.download("extracao_textos_artigos_pdfs.csv")
```
